```python
# Q1
df1 = pd.DataFrame(st1)
df1
```

```python
# Q2
df2 = pd.DataFrame(st2)
df2
```

```python
# Q3
df = pd.merge(df1,df2,left_on = "Stu_ID",right_on = "S_ID",how = "inner")
df
```

```python
# Q4
df.info()
```

```python
# Q5
df.isnull().sum()
```

```python
# Q6
# a
df['Co_Curr_Score'] = df['Co_Curr_Score'].fillna(df['Co_Curr_Score'].mean())
# b
df['Personality_Score'] = df['Personality_Score'].fillna(df['Personality_Score'].mean(),inplace = True)
```

```python
# c
df['Fee'] = df['Fee'].fillna(df['Fee'].mode())
df['Opted_Bus'] = df['Opted_Bus'].fillna(df['Opted_Bus'].mode())
```

```python
# d
df.drop('CReg_No',axis=1,inplace=True)
```

```python
# 9 
df['Total_Followers'] = pd.to_numeric(df['Total_Followers'])
df['DOB'] = pd.to_datetime(df['DOB'])
```

```python
# 10
# Create Year_Birth column
df['Year_Birth'] = df['DOB'].dt.year

# Create Quarter_Birth column
df['Quarter_Birth'] = df['DOB'].dt.quarter

# Create Month_Birth column (Month Name)
df['Month_Birth'] = df['DOB'].dt.month_name()
```

```python
# 15 Find Name,DOB,City,Math_Score,Active_On for students where students have scored between 75 to 85 in Math_Score or students are active on FB or IG.
df.loc[
    (df['Math_Score'].between(75,85)) | ((df['Active_On'] == 'FB') | (df['Active_On'] == 'IG')),
    ['Name','DOB','City','Math_Score','Active_On']
]
```

```python
# 14 Find Name,DOB,City,Math_Score,Personality Score for students 
# where students are not born in Mar, Oct or Nov or have Score at least 4.0 in Personality Score
df.loc[
    (
        (df['Month_Birth'] != 'March') &
        (df['Month_Birth'] != 'October') &
        (df['Month_Birth'] != 'November')
    ) |
    (df['Personality_Score'] >= 4.0),
    ['Name', 'DOB', 'City', 'Math_Score', 'Personality_Score']
]
```
